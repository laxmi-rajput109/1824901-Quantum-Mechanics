```python
import numpy as np

def radial_system(X, r, energy, potential, C, l):
    u = X[0]
    du = X[1]
    V = potential(r)
    d2u = ((V-energy)/C + l*(l+1)/r**2)*u
    return np.array([du, d2u])
```

```python
def radial_initial_conditions(r_min, l):
    u = r_min**(l+1)
    du = (l+1)*r_min**l
    return np.array([u, du])
```

```python
def radial_residual(r_min, r_max, h, energy, potential, C, l):

    N = int(np.ceil((r_max-r_min)/h))+1

    r = np.linspace(r_min, r_max, N)
    h = r[1]-r[0]

    X = np.zeros((N,2))

    X[0] = radial_initial_conditions(r_min, l)

    for i in range(N-1):

        k1 = h*radial_system(X[i], r[i], energy, potential, C, l)

        k2 = h*radial_system(X[i]+k1/2, r[i]+h/2, energy, potential, C, l)

        k3 = h*radial_system(X[i]+k2/2, r[i]+h/2, energy, potential, C, l)

        k4 = h*radial_system(X[i]+k3, r[i]+h, energy, potential, C, l)

        X[i+1] = X[i]+(k1+2*k2+2*k3+k4)/6

    R = X[-1,0]

    return X, r, R
```
